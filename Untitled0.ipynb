```python
import time, warnings
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.compose import ColumnTransformer
from sklearn.ensemble import RandomForestClassifier
from sklearn.impute import SimpleImputer
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score, precision_score, recall_score
from sklearn.metrics import f1_score, roc_auc_score, confusion_matrix
from sklearn.model_selection import GridSearchCV, train_test_split
from sklearn.neighbors import KNeighborsClassifier
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import OneHotEncoder, StandardScaler
from sklearn.svm import SVC
from sklearn.tree import DecisionTreeClassifier

warnings.filterwarnings("ignore")
sns.set_theme(style="whitegrid")
SEED = 42
CSV = "WA_Fn-UseC_-Telco-Customer-Churn.csv"

# ===== PART 1: DATA PREPROCESSING =====
df = pd.read_csv(CSV)
print("Loaded:", df.shape)
df["TotalCharges"] = pd.to_numeric(df["TotalCharges"], errors="coerce")
print("Missing values:")
print(df.isna().sum()[df.isna().sum() > 0])

cols_no_id = [c for c in df.columns if c != "customerID"]
n_before = len(df)
df = df.drop_duplicates(subset=cols_no_id)
print("Duplicates removed:", n_before - len(df))

df = df.drop(columns=["customerID"])
df["Churn"] = df["Churn"].map({"Yes": 1, "No": 0})

# feature engineering
svc = ["PhoneService", "OnlineSecurity", "OnlineBackup", "DeviceProtection",
       "TechSupport", "StreamingTV", "StreamingMovies"]
df["NumServices"] = (df[svc] == "Yes").sum(axis=1)
df["AvgMonthlySpend"] = df["TotalCharges"] / df["tenure"].replace(0, np.nan)
df["TenureGroup"] = pd.cut(df["tenure"], bins=[-1, 12, 24, 48, 72],
                           labels=["0-1y", "1-2y", "2-4y", "4-6y"]).astype(str)

# quick EDA
fig, ax = plt.subplots(1, 3, figsize=(16, 4))
df["Churn"].value_counts().plot.bar(ax=ax[0], color=["#4C72B0", "#C44E52"])
ax[0].set_title("Class balance (0=stayed, 1=churned)")
sns.boxplot(data=df, x="Churn", y="MonthlyCharges", ax=ax[1])
ax[1].set_title("Monthly charges vs churn")
sns.barplot(data=df, x="Contract", y="Churn", ax=ax[2])
ax[2].set_title("Churn rate by contract")
plt.tight_layout()
plt.savefig("eda_overview.png", dpi=120)
plt.show()

plt.figure(figsize=(7, 5))
sns.heatmap(df.select_dtypes("number").corr(), annot=True, fmt=".2f", cmap="coolwarm")
plt.title("Correlation heatmap")
plt.tight_layout()
plt.savefig("correlation_heatmap.png", dpi=120)
plt.show()

# split
X = df.drop(columns="Churn")
y = df["Churn"]
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.2, stratify=y, random_state=SEED)
print("Train:", X_train.shape, "Test:", X_test.shape)

# encode + scale (fit on train only, inside pipeline)
num_cols = X.select_dtypes(include="number").columns.tolist()
cat_cols = X.select_dtypes(exclude="number").columns.tolist()
prep = ColumnTransformer([
    ("num", Pipeline([("imp", SimpleImputer(strategy="median")),
                      ("sc", StandardScaler())]), num_cols),
    ("cat", Pipeline([("imp", SimpleImputer(strategy="most_frequent")),
                      ("oh", OneHotEncoder(handle_unknown="ignore"))]), cat_cols),
])

# ===== PART 2 + 3: TRAIN AND EVALUATE =====
models = {
    "Logistic Regression": LogisticRegression(max_iter=1000, random_state=SEED),
    "Decision Tree": DecisionTreeClassifier(max_depth=6, random_state=SEED),
    "Random Forest": RandomForestClassifier(n_estimators=300, random_state=SEED, n_jobs=-1),
    "KNN": KNeighborsClassifier(n_neighbors=15),
    "SVM (RBF)": SVC(probability=True, random_state=SEED),
}


def evaluate(name, pipe, secs):
    pred = pipe.predict(X_test)
    prob = pipe.predict_proba(X_test)[:, 1]
    row = {
        "Model": name,
        "Training Time (s)": round(secs, 3),
        "Accuracy": accuracy_score(y_test, pred),
        "Precision": precision_score(y_test, pred),
        "Recall": recall_score(y_test, pred),
        "F1 Score": f1_score(y_test, pred),
        "ROC-AUC": roc_auc_score(y_test, prob),
    }
    return row, confusion_matrix(y_test, pred)


results = []
cms = {}
fitted = {}
for name, clf in models.items():
    pipe = Pipeline([("prep", prep), ("model", clf)])
    t0 = time.perf_counter()
    pipe.fit(X_train, y_train)
    secs = time.perf_counter() - t0
    row, cm = evaluate(name, pipe, secs)
    results.append(row)
    cms[name] = cm
    fitted[name] = pipe
    print("Trained:", name)

# confusion matrices
fig, axes = plt.subplots(2, 3, figsize=(15, 8))
flat = axes.ravel()
for i, (name, cm) in enumerate(cms.items()):
    sns.heatmap(cm, annot=True, fmt="d", cmap="Blues", cbar=False, ax=flat[i],
                xticklabels=["Stay", "Churn"], yticklabels=["Stay", "Churn"])
    flat[i].set_title(name)
    flat[i].set_xlabel("Predicted")
    flat[i].set_ylabel("Actual")
for j in range(len(cms), len(flat)):
    flat[j].axis("off")
plt.tight_layout()
plt.savefig("confusion_matrices.png", dpi=120)
plt.show()

# ===== PART 4: COMPARISON TABLE =====
pros = {
    "Logistic Regression": "Fast, interpretable, good probabilities, strong baseline",
    "Decision Tree": "Easy to explain, handles non-linear rules, no scaling needed",
    "Random Forest": "Robust to noise, captures interactions, feature importance",
    "KNN": "Simple, no training phase, adapts to local patterns",
    "SVM (RBF)": "Handles non-linear boundaries, good in high dimensions",
}
cons = {
    "Logistic Regression": "Assumes linear relationship with log-odds",
    "Decision Tree": "Overfits easily, unstable",
    "Random Forest": "Slower, less interpretable",
    "KNN": "Slow predictions, sensitive to scaling/irrelevant features",
    "SVM (RBF)": "Slow on large data, hard to interpret",
}
comp = pd.DataFrame(results).set_index("Model")
comp["Advantages"] = comp.index.map(pros)
comp["Limitations"] = comp.index.map(cons)
comp = comp.sort_values("F1 Score", ascending=False)
metric_cols = ["Training Time (s)", "Accuracy", "Precision", "Recall", "F1 Score", "ROC-AUC"]
print("\n===== MODEL COMPARISON =====")
print(comp[metric_cols].round(3).to_string())
comp.round(4).to_csv("model_comparison.csv")

comp[["Accuracy", "Precision", "Recall", "F1 Score", "ROC-AUC"]].plot.bar(
    figsize=(11, 5), rot=20)
plt.title("Model comparison (test set)")
plt.ylim(0, 1)
plt.legend(loc="lower right", ncol=5)
plt.tight_layout()
plt.savefig("model_comparison.png", dpi=120)
plt.show()

# ===== BONUS: HYPERPARAMETER TUNING =====
print("\n===== TUNING (GridSearchCV) =====")
grids = {
    "Logistic Regression": (
        LogisticRegression(max_iter=1000, random_state=SEED),
        {"model__C": [0.01, 0.1, 1, 10],
         "model__class_weight": [None, "balanced"]}),
    "Random Forest": (
        RandomForestClassifier(random_state=SEED, n_jobs=-1),
        {"model__n_estimators": [200, 400],
         "model__max_depth": [6, 10, None],
         "model__min_samples_leaf": [1, 3],
         "model__class_weight": [None, "balanced"]}),
}
tuned = []
for name, (clf, grid) in grids.items():
    pipe = Pipeline([("prep", prep), ("model", clf)])
    gs = GridSearchCV(pipe, grid, scoring="f1", cv=5, n_jobs=-1)
    t0 = time.perf_counter()
    gs.fit(X_train, y_train)
    secs = time.perf_counter() - t0
    row, cm = evaluate(name + " (tuned)", gs.best_estimator_, secs)
    tuned.append(row)
    fitted[name + " (tuned)"] = gs.best_estimator_
    print(name, "| best CV F1:", round(gs.best_score_, 3))
    print("   best params:", gs.best_params_)

final = pd.concat([comp[metric_cols],
                   pd.DataFrame(tuned).set_index("Model")[metric_cols]])
final = final.sort_values("F1 Score", ascending=False)
print("\n===== FINAL RANKING (baseline + tuned) =====")
print(final.round(3).to_string())
final.round(4).to_csv("final_ranking.csv")

# ===== PART 5: CONCLUSION =====
best_name = final.index[0]
best = final.iloc[0]
print("\n===== CONCLUSION =====")
print("Best model:", best_name)
print(best.round(3).to_string())

best_pipe = fitted[best_name]
names = best_pipe.named_steps["prep"].get_feature_names_out()
est = best_pipe.named_steps["model"]
if hasattr(est, "feature_importances_"):
    imp = pd.Series(est.feature_importances_, index=names)
elif hasattr(est, "coef_"):
    imp = pd.Series(est.coef_[0], index=names)
else:
    imp = None
if imp is not None:
    top = imp.reindex(imp.abs().sort_values(ascending=False).index).head(10)
    print("\nTop 10 features:")
    print(top.round(3).to_string())
    top[::-1].plot.barh(figsize=(8, 5), color="#4C72B0")
    plt.title("Top features - " + best_name)
    plt.tight_layout()
    plt.savefig("feature_importance.png", dpi=120)
    plt.show()

print("\nDone. Files saved in the Colab file panel (left side).")
```
